## EXEMPLO COMPLETO: SCORING EM BATCH PARA MÚLTIPLOS TICKERS"

```python
print("\n" + "=" * 60)
print("9. EXEMPLO COMPLETO: SCORING EM BATCH PARA MÚLTIPLOS TICKERS")
print("=" * 60)

"""
EXEMPLO DE SCORING EM BATCH (DOCUMENTAÇÃO - NÃO EXECUTAR)

Este exemplo mostra como escorar múltiplos tickers de uma vez:

    from simple_quant import db, features, models
    import pandas as pd
    from pathlib import Path

    # 1. Carregar todos os modelos disponíveis
    modelos_df = db.list_models()
    print(f"Modelos disponíveis: {len(modelos_df)}")

    # 2. Carregar base completa
    df_completo = db.load_dataset()

    # 3. Para cada modelo, escorar seu ticker correspondente
    for _, modelo in modelos_df.iterrows():
        ticker = modelo['ticker']
        model_name = modelo['model_name']

        print(f"\\nProcessando {model_name} para {ticker}...")

        try:
            # Filtrar dados do ticker
            df_ticker = df_completo[df_completo['ticker'] == ticker]

            if len(df_ticker) == 0:
                print(f"  Sem dados para {ticker}")
                continue

            # Preparar dados de modelagem
            df_modeling = features.prepare_modeling_data(df_ticker, ticker)

            # Carregar modelo
            model, metadata = models.load_model_artifacts(modelo['artifact_path'])

            # Pegar features do modelo
            feature_cols = modelo['features']

            # Para cada linha disponível, fazer scoring
            for idx in range(len(df_modeling)):
                # Preparar linha específica para scoring
                X_row, row_data = features.get_scoring_row(
                    df_model=df_modeling,
                    feature_cols=feature_cols,
                    as_of_date=df_modeling.iloc[idx]['data'].strftime('%Y-%m-%d')
                )

                # Executar scoring
                result = models.score_row(model, X_row, modelo['threshold'])

                # Salvar score
                db.save_score_row(
                    model_name=model_name,
                    ticker=ticker,
                    target_col=modelo['target_col'],
                    probability=result['probability'],
                    prediction=result['prediction'],
                    as_of_date=row_data['data'].strftime('%Y-%m-%d')
                )

            print(f"  ✓ {len(df_modeling)} scores salvos para {ticker}")

        except Exception as e:
            print(f"  ✗ Erro processando {model_name}: {e}")

    print("\\nBatch scoring concluído!")

Este processo pode ser automatizado para rodar diariamente,
escorando todos os modelos para todos os tickers disponíveis.
"""
```
